**Problema:**

Seja $V=\{(w,x,y,z)\big |\, w,x,y,z>0,\,w,x,y,z\in\mathbb{R}\}$, em que o produto por escalar $tu$ e a soma de vetores $a+b$, com $a,b\in V$ e $t\in \mathbb{R}$, são definidos como:

*   $ta=(w^t,x^t,y^t,z^t)$, para $a=(w,x,y,z)$;
*   $a+b=(we,xf,yg,zh)$, para $b=(e,f,g,h)$;
*   A regra de prioridade das operações é primeiro multiplicar pelo escalar e depois somar.

Seja $f:V\times V\to \mathbb{R}$ o produto interno definido sobre $V$.

Seja $T:\mathbb{R}^3\to V$ a transformação linear, dada por $$T(r)=\alpha v_1+\beta v_2+\gamma v_3,\qquad r=(\alpha,\beta,\gamma).$$

Sejam $v_1=(1,2,4,8)$, $v_2=(2,1,8,8)$, $v_3=(4,4,2,2)$ e $k=(1,4,2,4)$ em $V$.

Sejam $A=\{(2,1,1,1),(1,2,1,1),(1,1,2,1),(1,1,1,2)\}$ e $B=\{v_1,v_2,v_3,k\}$ e $C=\{(2,1,1,1),(2,2,1,1),(2,2,2,1),(2,2,2,2)\}$ bases de $V$.

a) Determine a matriz de mudança da base $B$ para a base $C$ e a matriz de mudança da base $B$ para a base $A$. Escreva as coordenadas de $v_1,v_2,v_3,k$, na base $A$ e na base $C$.

b) Ortogonalize a base $\mathcal{B}$, considerando $$f(a,b)=(a_1,a_2,a_3,a_4)\cdot (b_1,b_2,b_3,b_4),$$ sendo $(a_1,a_2,a_3,a_4),\, (b_1,b_2,b_3,b_4)$ as coordenadas de $a$ e $b$, na base $\mathcal{A}$.

c) Resolver por meio do processo de ortogonalização a equação $$T(r)=k,$$ considerando $$f(a,b)=(a_1,a_2,a_3,a_4)\cdot (b_1,b_2,b_3,b_4),$$ sendo $(a_1,a_2,a_3,a_4),\, (b_1,b_2,b_3,b_4)$ as coordenadas de $a$ e $b$, na base $\mathcal{A}$. Determine o erro e o módulo do erro cometido.

d) Resolver por meio do processo de ortogonalização a equação $$T(r)=k,$$ considerando $$f(a,b)=(a_1,a_2,a_3,a_4)\cdot (b_1,b_2,b_3,b_4),$$ sendo $(a_1,a_2,a_3,a_4),\, (b_1,b_2,b_3,b_4)$ as coordenadas de $a$ e $b$, na base $\mathcal{C}$. Determine o erro e o módulo do erro cometido.
